```python
import copy
import nglview as nv

from porems import *
```

```python
def show_mol(mol):
    temp = copy.deepcopy(mol)
    
    temp.rotate("x",-90)
    temp.zero()

    Store(temp, "output").pdb()
    return nv.show_structure_file("output/Molecule.pdb", gui=True)
```

# Create Molecule
[Website](https://calculla.com/bond_lengths) for bond-lengths.

* C=O : 120.2 pm
* C-O : 134.3 pm
* C-H : 109.7 pm
* O-H : 97.2 pm

```python
mol = Molecule()
    
mol.add("C", [0, 0, 0])
mol.add("O", 0, r=0.1202)
mol.add("O", 0, r=0.1343, theta= 120)
mol.add("H", 0, r=0.1097, theta=-120)
mol.add("H", 2, r=0.0972, theta=  60)

mol.zero()        

# Show molecule
view = show_mol(mol)

mol.set_name("formic_acid")
mol.set_short("1ooh")

# Save molecule
Store(mol).gro()
Store(mol).pdb()

display(view)
```
