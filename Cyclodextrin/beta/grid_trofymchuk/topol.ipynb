```python
def read(link):
    # Initialize
    data = {}

    # Read data
    is_new = False
    block = ""
    with open(link, "r") as file_in:
        for line in file_in:
            dat = line.split()

            if is_new:
                if len(dat) > 0:
                    data[block].append(line)
                else:
                    is_new = False

            elif len(dat) > 0 and dat[0] == "[":
                is_new = True
                block = dat[1]
                data[block] = []

    return data
```

```python
data = read("../native/cd_BCD.itp")

for x in data: print(x)
```

```python
atoms = data["atoms"]
atoms_data = [atom.split() for atom in atoms]

# Remove hydrogen
del_list = []
for i in range(1,len(atoms_data)):
    if int(atoms_data[i][0]) == 10:
        del_list.append(i)
        
for x in reversed(sorted(del_list)):
    atoms_data.pop(x)

# Run through atoms
for atom in atoms_data[1:]:
    # Update index
    idx = int(atom[0])
    idx_new = str(idx+14) if idx <= 9 else str(idx+13)
    atom[0] = idx_new
    atom[5] = idx_new
    
    # Update residue index
    atom[2] = str(int(atom[2])+1)
        
# Test
print(atoms[1])
for atom in atoms_data[1:15]:
    print("%6s%11s%7s%7s%7s%7s%11s%11s"%tuple(atom))
```

```python
bonds = data["bonds"]
bonds_data = [bond.split() for bond in bonds]

# Remove hydrogen
del_list = []
for i in range(1,len(bonds_data)):
    if int(bonds_data[i][0]) == 10 or int(bonds_data[i][1]) == 10:
        del_list.append(i)

for x in reversed(sorted(del_list)):
    bonds_data.pop(x)

# Run through bonds
for bond in bonds_data[1:]:
    # Update index
    idx_a = int(bond[0])
    idx_b = int(bond[1])
    idx_a_new = str(idx_a+14) if idx_a <= 9 else str(idx_a+13)
    idx_b_new = str(idx_b+14) if idx_b <= 9 else str(idx_b+13)
    bond[0] = idx_a_new
    bond[1] = idx_b_new
        
# Test
print(bonds[1])
for bond in bonds_data[1:10]:
    print("%5s%6s%6s%12s%12s"%tuple(bond))
```

```python
pairs = data["pairs"]
pairs_data = [pair.split() for pair in pairs]

# Remove hydrogen
del_list = []
for i in range(1,len(pairs_data)):
    if int(pairs_data[i][0]) == 10 or int(pairs_data[i][1]) == 10:
        del_list.append(i)

for x in reversed(sorted(del_list)):
    pairs_data.pop(x)

# Run through pairs
for pair in pairs_data[1:]:
    # Update index
    idx_a = int(pair[0])
    idx_b = int(pair[1])
    idx_a_new = str(idx_a+14) if idx_a <= 9 else str(idx_a+13)
    idx_b_new = str(idx_b+14) if idx_b <= 9 else str(idx_b+13)
    pair[0] = idx_a_new
    pair[1] = idx_b_new
        
# Test
print(pairs[1])
for pair in pairs_data[1:10]:
    print("%6s%7s%7s"%tuple(pair))
```

```python
angles = data["angles"]
angles_data = [angle.split() for angle in angles]

# Remove hydrogen
del_list = []
for i in range(1,len(angles_data)):
    if int(angles_data[i][0]) == 10 or int(angles_data[i][1]) == 10 or int(angles_data[i][2]) == 10:
        del_list.append(i)

for x in reversed(sorted(del_list)):
    angles_data.pop(x)

# Run through angles
for angle in angles_data[1:]:
    # Update index
    idx_a = int(angle[0])
    idx_b = int(angle[1])
    idx_c = int(angle[2])
    idx_a_new = str(idx_a+14) if idx_a <= 9 else str(idx_a+13)
    idx_b_new = str(idx_b+14) if idx_b <= 9 else str(idx_b+13)
    idx_c_new = str(idx_c+14) if idx_c <= 9 else str(idx_c+13)
    angle[0] = idx_a_new
    angle[1] = idx_b_new
    angle[2] = idx_c_new
        
# Test
print(angles[1])
for angle in angles_data[1:10]:
    print("%5s%6s%6s%6s%12s%12s"%tuple(angle))
```

```python
dihedrals = data["dihedrals"]
dihedrals_data = [dihedral.split() for dihedral in dihedrals]

# Remove hydrogen
del_list = []
for i in range(1,len(dihedrals_data)):
    if int(dihedrals_data[i][0]) == 10 or int(dihedrals_data[i][1]) == 10 or int(dihedrals_data[i][2]) == 10 or int(dihedrals_data[i][3]) == 10:
        del_list.append(i)

for x in reversed(sorted(del_list)):
    dihedrals_data.pop(x)

# Run through dihedrals
for dihedral in dihedrals_data[1:]:
    # Update index
    idx_a = int(dihedral[0])
    idx_b = int(dihedral[1])
    idx_c = int(dihedral[2])
    idx_d = int(dihedral[3])
    idx_a_new = str(idx_a+14) if idx_a <= 9 else str(idx_a+13)
    idx_b_new = str(idx_b+14) if idx_b <= 9 else str(idx_b+13)
    idx_c_new = str(idx_c+14) if idx_c <= 9 else str(idx_c+13)
    idx_d_new = str(idx_d+14) if idx_d <= 9 else str(idx_d+13)
    dihedral[0] = idx_a_new
    dihedral[1] = idx_b_new
    dihedral[2] = idx_c_new
    dihedral[3] = idx_d_new
        
# Test
print(dihedrals[1])
for dihedral in dihedrals_data[1:10]:
    print("%6s%5s%5s%5s%6s%12s%12s%12s%12s%12s%12s%3s"%tuple(dihedral))
```

```python
with open("bcd_new.top", "w") as file_out:
    # Defaults
    file_out.write("[ "+"defaults"+" ]\n")
    for line in data["defaults"]:
        file_out.write(line)
    file_out.write("\n")
    
    # Atomtypes
    file_out.write("[ atomtypes ]\n")
    for line in data["atomtypes"]:
        file_out.write(line)
    file_out.write("\n")
    
    # Moleculetype
    file_out.write("[ moleculetype ]\n")
    for line in data["moleculetype"]:
        file_out.write(line)
    file_out.write("\n")

    # Atoms
    file_out.write("[ atoms ]\n")
    file_out.write(atoms[0])
    for atom in atoms_data[1:]:
        file_out.write("%6s%11s%7s%7s%7s%7s%11s%11s"%tuple(atom)+"\n")
    file_out.write("\n")
    
    # Bonds
    file_out.write("[ bonds ]\n")
    file_out.write(bonds[0])
    for bond in bonds_data[1:]:
        file_out.write("%5s%6s%6s%12s%12s"%tuple(bond)+"\n")
    file_out.write("\n")
    
    # Pairs
    file_out.write("[ pairs ]\n")
    file_out.write(pairs[0])
    for pair in pairs_data[1:]:
        file_out.write("%6s%7s%7s"%tuple(pair)+"\n")
    file_out.write("\n")
    
    # Angles
    file_out.write("[ angles ]\n")
    file_out.write(angles[0])
    for angle in angles_data[1:]:
        file_out.write("%5s%6s%6s%6s%12s%12s"%tuple(angle)+"\n")
    file_out.write("\n")
    
    # Dihedrals
    file_out.write("[ dihedrals ]\n")
    file_out.write(dihedrals[0])
    for dihedral in dihedrals_data[1:]:
        file_out.write("%6s%5s%5s%5s%6s%12s%12s%12s%12s%12s%12s%3s"%tuple(dihedral)+"\n")
    file_out.write("\n")
    
```
