```python
import copy
import nglview as nv

from porems import *
```

```python
def show_mol(mol):
    temp = copy.deepcopy(mol)
    
    temp.rotate("x",-90)
    temp.zero()

    Store(temp,"output").pdb()
    view = nv.show_structure_file("output/Molecule.pdb", gui=True)
    
    display(view)
```

# Create Tail-Structure
Website for bond-lengths [link](https://www.chem.tamu.edu/rgroup/connell/linkfiles/bonds.pdf).

```python
connect = Molecule()
is_si = True

if is_si:
    connect.add("Si", [0, 0, 0])
    connect.add("C", 0, r=0.185)
else:
    connect.add("Ci", [0, 0, 0])
    connect.add("C", 0, r=0.154)
    
connect.add("C", 1, r=0.154, theta=30)
connect.add("C", 2, r=0.154, theta=-30)
connect.add("N", 3, r=0.147, theta=30)
connect.add("C", 4, r=0.147, theta=-30)

connect.add("O", 5, r=0.124, theta=60)

connect.add("H", 1, r=0.109, theta=-90, phi=60)
connect.add("H", 1, r=0.109, theta=90, phi=120)
connect.add("H", 2, r=0.109, theta=90, phi=60)
connect.add("H", 2, r=0.109, theta=-90, phi=120)
connect.add("H", 3, r=0.109, theta=-90, phi=60)
connect.add("H", 3, r=0.109, theta=90, phi=120)

connect.add("H", 4, r=0.101, theta=90)

connect.add("O", 5, r=0.143, theta=-60)

# Glucopyranose
connect.add("C", 14, r=0.143, theta=-90)

connect.add("C", 15, r=0.143, theta=-90)
connect.add("O", 16, r=0.150, theta=-30)
connect.add("C", 16, r=0.150, theta=-150)
connect.add("C", 17, r=0.150, theta=-90)
connect.add("C", 18, r=0.150, theta=-90)
connect.add("C", 20, r=0.150, theta=-30)

connect.add("O", 18, r=0.143, theta=120, phi=-60)
connect.add("O", 19, r=0.143, theta=-50, phi=60)
connect.add("O", 20, r=0.143, theta=240, phi=-60)
connect.add("O", 21, r=0.143, theta=-90, phi=60)

connect.add("H", 22, r=0.096, theta=150)
connect.add("H", 18, r=0.109, theta=120, phi=60)
connect.add("H", 23, r=0.096, theta=-30)
connect.add("H", 19, r=0.109, theta=-50, phi=-60)
connect.add("H", 24, r=0.096, theta=210)
connect.add("H", 20, r=0.109, theta=240, phi=60)
connect.add("H", 25, r=0.096, theta=-90)
connect.add("H", 21, r=0.109, theta=-90, phi=-60)
connect.add("H", 16, r=0.109, theta=-90, phi=90)
connect.add("H", 15, r=0.109, theta=-60, phi=-90)
connect.add("H", 15, r=0.109, theta=-60, phi=90)

if not is_si:
    for i in range(3):
        connect.add("H", 0, r=0.109, theta=-120, phi=60+120*i)

# Show molecule
show_mol(connect)

# Save molecule
connect.set_name("cydtail")
connect.set_short("CYT")

Store(connect, "connect_tail").gro()
Store(connect, "connect_tail").pdb()
Store(connect, "connect_tail").job()
```
