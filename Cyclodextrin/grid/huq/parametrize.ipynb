```python
import copy
import nglview as nv

from porems import *
```

```python
def show_mol(mol):
    temp = copy.deepcopy(mol)
    
    temp.rotate("x",-90)
    temp.zero()

    Store(temp,"output").pdb()
    view = nv.show_structure_file("output/Molecule.pdb", gui=True)
    
    display(view)
```

# Create Tail-Structure
Website for bond-lengths [link](https://www.chem.tamu.edu/rgroup/connell/linkfiles/bonds.pdf).

```python
tail = Molecule()
is_si = True

if is_si:
    tail.add("Si", [0, 0, 0])
    tail.add("C", 0, r=0.185)
else:
    tail.add("Ci", [0, 0, 0])
    tail.add("C", 0, r=0.154)
    
tail.add("C", 1, r=0.154, theta=30)
tail.add("C", 2, r=0.154, theta=-30)
tail.add("N", 3, r=0.147, theta=30)
tail.add("C", 4, r=0.147, theta=-30)
tail.add("N", 5, r=0.129, theta=-60)
tail.add("N", 5, r=0.129, theta=60)
tail.add("C", 6, r=0.129)
tail.add("C", 7, r=0.129)
tail.add("N", 9, r=0.129, theta=-60)

tail.add("O", 9, r=0.143, theta=60)
tail.add("O", 8, r=0.143, theta=-60)

tail.add("H", 1, r=0.109, theta=-90, phi=60)
tail.add("H", 1, r=0.109, theta=90, phi=120)
tail.add("H", 2, r=0.109, theta=90, phi=60)
tail.add("H", 2, r=0.109, theta=-90, phi=120)
tail.add("H", 3, r=0.109, theta=-90, phi=60)
tail.add("H", 3, r=0.109, theta=90, phi=120)

tail.add("H", 4, r=0.101, theta=90)

tail.add("H", 11, r=0.096, theta=90)

if not is_si:
    tail.add("H", 12, r=0.096, theta=-90)
    
    for i in range(3):
        tail.add("H", 0, r=0.109, theta=-120, phi=60+120*i)

# Show molecule
show_mol(tail)

# Save molecule
tail.set_name("cydtail")
tail.set_short("CYT")

Store(tail,"tail").gro()
Store(tail,"tail").pdb()
Store(tail,"tail").job()
```
