```python
def show_mol(mol):
    temp = copy.deepcopy(mol)
    temp.set_box(Molecule(inp=temp.get_mol_list()).get_box())
    #print(Molecule(inp=temp.get_mol_list()).get_box())
    
    for mol in temp.get_mol_list():
        mol.rotate("x",-90)
        #mol.zero()
                 
    Store(temp,"output").pdb()
    view = nv.show_structure_file("output/Molecule.pdb", gui=False)
    
    display(view)
```
