```python
import copy
import nglview as nv

from porems import *
```

```python
def show_mol(mol):
    temp = copy.deepcopy(mol)
    
    temp.rotate("x",-90)
    temp.zero()

    Store(temp,"output").pdb()
    view = nv.show_structure_file("output/Molecule.pdb", gui=True)
    
    display(view)
```

# Create Tail-Structure
[Website](https://www.chem.tamu.edu/rgroup/connell/linkfiles/bonds.pdf) for bond-lengths.

```python
mol = Molecule()
is_si = True

if is_si:
    mol.add("Si", [0, 0, 0])
    mol.add("C", 0, r=0.185)
else:
    mol.add("Ci", [0, 0, 0])
    mol.add("C", 0, r=0.154)
    
mol.add("C", 1, r=0.154, theta=30)
mol.add("C", 2, r=0.154, theta=-30)
mol.add("N", 3, r=0.147, theta=30)

mol.add("H", 1, r=0.109, theta=-90, phi=60)
mol.add("H", 1, r=0.109, theta=90, phi=120)
mol.add("H", 2, r=0.109, theta=90, phi=60)
mol.add("H", 2, r=0.109, theta=-90, phi=120)
mol.add("H", 3, r=0.109, theta=-90, phi=60)
mol.add("H", 3, r=0.109, theta=90, phi=120)
mol.add("H", 4, r=0.109, theta=90, phi=60)
mol.add("H", 4, r=0.109, theta=-90, phi=120)

if not is_si:
    for i in range(3):
        mol.add("H", 0, r=0.109, theta=-120, phi=60+120*i)

# Show molecule
show_mol(mol)

# Save molecule
mol.set_name("propylamine")
mol.set_short("PPA")

Store(mol).gro()
Store(mol).pdb()
Store(mol).job()
```
