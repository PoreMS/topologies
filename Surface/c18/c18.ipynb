```python
import copy
import nglview as nv

from porems import *
```

```python
def show_mol(mol):
    temp = copy.deepcopy(mol)
    
    temp.rotate("x",-90)
    temp.zero()

    Store(temp, "output").pdb()
    return nv.show_structure_file("output/Molecule.pdb", gui=True)
```

# Create Tail-Structure
[Website](https://calculla.com/bond_lengths) for bond-lengths.

```python
mol = Molecule()
is_si = True

if is_si:
    mol.add("Si", [0, 0, 0])
    mol.add("O",  0, r=0.155)
    mol.add("Si", 1, r=0.155)
    mol.add("C",  2, r=0.190, phi= 60, theta=90)
    mol.add("C",  2, r=0.190, phi=-60, theta=90)
    mol.add("C",  2, r=0.190)
else:
    mol.add("C", [0, 0, 0])
    mol.add("O", 0, r=0.135)
    mol.add("C", 1, r=0.135)
    mol.add("C", 2, r=0.155, phi= 60, theta=90)
    mol.add("C", 2, r=0.155, phi=-60, theta=90)
    mol.add("C", 2, r=0.155)
    
for i in range(18-1):
    mol.add("C", 5+i, r=0.155)
    
for atom_id in [3, 4]:
    for j in range(3):
        mol.add("H", atom_id, r=0.109, theta=60, phi=60+120*j)
        
for atom_id in range(5, 22):
    for j in range(2):
        mol.add("H", atom_id, r=0.109, theta=60, phi=-60+120*j)     

for j in range(3):
    mol.add("H", 22, r=0.109, theta=60, phi=60+120*j)

if not is_si:
    for i in range(3):
        mol.add("H", 0, r=0.109, theta=-120, phi=60+120*i)

mol.zero()        

# Show molecule
view = show_mol(mol)

mol.set_name("c18")
mol.set_short("C18")

# Save molecule
Store(mol).gro()
Store(mol).pdb()

display(view)
```
